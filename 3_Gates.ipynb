# Linear transformations
* A transformation $T : V\to V$ is linear if 
        $$T(a\ket{\psi} + b\ket{\phi}) = aT(\ket{\psi}) + bT(\ket{\phi})$$
* For any state $\ket{\psi} = a\ket{0} + b\ket{1}$, we have
$$T(\ket{\psi}) = aT(\ket{0}) + bT(\ket{1})$$
* Hence it is sufficient to specify the action of $T$ on the basis vectors.
* For example, consider a transformation $H$ defined as $$
        \begin{aligned}
        H(\ket{0}) &= \frac{\ket{0} + \ket{1}}{\sqrt{2}} = \ket{+} \\
        H(\ket{1}) &= \frac{\ket{0} - \ket{1}}{\sqrt{2}} = \ket{-}
        \end{aligned}
        $$
* In quantum mechanics, transformations are called **operators**.
* We often write $T(\ket{\psi})$ as $T\ket{\psi}$ and say that operator $T$ is acting on ket $\ket{\psi}$.

## Writing an operator in outer product notation
* Suppose an operator acts on a basis as follows $$
        \begin{aligned}
        T\ket{0} &= \ket{\psi_0} \\
        T\ket{1} &= \ket{\psi_1}
        \end{aligned}
        $$
* Then we can write $T$ as $$T = \outpro{\psi_0}{0} + \outpro{\psi_1}{1}$$
* Such that 
        $$\begin{aligned}
        T\ket{0} &= (\outpro{\psi_0}{0} + \outpro{\psi_1}{1})\ket{0} = \outpro{\psi_0}{0}\ket{0} + \outpro{\psi_1}{1}\ket{0} = \ket{\psi_0}\\
        T\ket{1} &= (\outpro{\psi_0}{0} + \outpro{\psi_1}{1})\ket{1} = \outpro{\psi_0}{0}\ket{1} + \outpro{\psi_1}{1}\ket{1} = \ket{\psi_1}
        \end{aligned}$$
* Alternatively, we can write 
$$
\begin{aligned}
T &= a\outpro{0}{0} + b\outpro{0}{1} + c\outpro{1}{0} + d\outpro{1}{1}\\ 
  &= \begin{bmatrix}
        a & b\\ 
        c & d 
      \end{bmatrix}
\end{aligned}$$

## Adjoint of an operator $T^\dagger$
* Recall dual of a ket is bra $\bra{\psi} = \ket{\psi}^\dagger$ (column $\to$ row with conjugation)
* Similarly, we can define dual (called adjoint) of an operator.
* But, first dual of an outer product needs to be defined which is $$\outpro{\psi}{\phi}^\dagger = \outpro{\phi}{\psi}$$
* Now adjoint of $T = a\outpro{0}{0} + b\outpro{0}{1} + c\outpro{1}{0} + d\outpro{1}{1}$ is defined as $$\begin{aligned} T^\dagger &= a^*\outpro{0}{0} + b^*\outpro{1}{0} + c^*\outpro{0}{1} + d^*\outpro{1}{1}\\ &= \begin{bmatrix}
        a^* & c^*\\ 
        b^* & d^* 
      \end{bmatrix}= (T^*)^T
\end{aligned}$$
* $T^\dagger$ is complex conjugate transpose of $T$.
* Adjoint of an operator is also an operator.

## $TT^\dagger \neq T^\dagger T$
* $$\begin{aligned}TT^\dagger = (|a|^2|+|b|^2)\outpro{0}{0} &+ (ac^*+bd^*)\outpro{0}{1}\\ + (|c|^2+|d|^2)\outpro{1}{1} &+ (a^*c+b^*d)\outpro{1}{0} \end{aligned}$$
* $$\begin{aligned}T^\dagger T = (|a|^2|+|c|^2)\outpro{0}{0} &+ (a^*b+c^*d)\outpro{0}{1}\\ + (|b|^2+|d|^2)\outpro{1}{1} &+ (ab^*+cd^*)\outpro{1}{0} \end{aligned}$$

# State evolution
* Since state is a vector, its change is achieved via transformations.
* Such transformations must be linear.
* Since we want vectors to be unit at each point of time, the state before the transform $U$ and state after the transform must have the same length. $$\ket{\psi(t)} = U(t)\ket{\psi(0)}$$
* In addition, we assume that evolution is time reversible.
* All these assumptions imply that $U$ must be unitary. $$UU^\dagger = U^\dagger = I$$
* In quantum computing, unitary operators are called **gates**. $\ket{\psi} = U\ket{\phi}$

## Single qubit Unitary operators
* As a single qubit state can be written as $\ket{\psi} = a \ket{0} + b\ket{1}$, action of single qubit gate can be seen as either modifying the coefficients $a$ and $b$ or states $\ket{0}$ and $\ket{1}$.
### Standard gates
* `XGate`: The single-qubit Pauli-X gate ($\sigma_x$)
    $$\begin{aligned}X\ket{0} &= \ket{1}\\ X\ket{1} &= \ket{0}\end{aligned}$$
    $$X =\begin{bmatrix}0 & 1\\1 & 0\end{bmatrix} = \outpro{0}{1} + \outpro{1}{0}$$
* `YGate`: The single-qubit Pauli-Y gate ($\sigma_y$) 
    $$\begin{aligned}Y\ket{0} &= -i\ket{1}\\ Y\ket{1} &= i\ket{0}\end{aligned}$$
    $$Y=i\begin{bmatrix}0 & -1\\1 & 0\end{bmatrix} = -i\outpro{0}{1} + i\outpro{1}{0}$$
* `ZGate`: The single-qubit Pauli-Z gate ($\sigma_z$)
    $$\begin{aligned}Z\ket{0} &= \ket{0}\\ Z\ket{1} &= -\ket{1}\end{aligned}$$
    $$Z=\begin{bmatrix}1 & 0\\0 & -1\end{bmatrix} = \outpro{0}{0} - \outpro{1}{1}$$
* `IGate`: The single-qubit Pauli-I gate ($I$)
    $$\begin{aligned}I\ket{0} &= \ket{0}\\ I\ket{1} &= \ket{1}\end{aligned}$$
    $$I=\begin{bmatrix}1 & 0\\0 & 1\end{bmatrix} = \outpro{0}{0} + \outpro{1}{1}$$
* `HGate`: The single-qubit Hadamard gate ($H$)
    $$\begin{aligned}H\ket{0} &= \frac{\ket{0}+\ket{1}}{\sqrt{2}}\\ H\ket{1} &= \frac{\ket{0}-\ket{1}}{\sqrt{2}}\end{aligned}$$
    $$H=\frac{1}{\sqrt{2}}\begin{bmatrix}1 & 1\\1 & -1\end{bmatrix} = \outpro{+}{0} + \outpro{-}{1}$$
* `SGate`: The single-qubit S-gate ($S$)
    $$\begin{aligned}S\ket{0} &= \ket{0}\\ S\ket{1} &= i\ket{1}\end{aligned}$$
    $$S=\begin{bmatrix}1 & 0\\0 & i\end{bmatrix} = \outpro{0}{0} + i\outpro{1}{1}$$
* `TGate`: The single-qubit T-gate ($T$)
    $$\begin{aligned}T\ket{0} &= \ket{0}\\ T\ket{1} &= e^{i\frac{\pi}{4}}\ket{1}\end{aligned}$$
    $$T=\begin{bmatrix}1 & 0\\0 & e^{i\frac{\pi}{4}}\end{bmatrix} = \outpro{0}{0} + e^{i\frac{\pi}{4}}\outpro{1}{1}$$

### Rotation gates (Parameterized gates)
* For a complex plane, $e^{i\theta}$ represents rotating a complex number by angle $\theta$.
* `RXGate (rx)` : Rotation about $x$-axis
    $$\begin{aligned}R_X(\theta)\ket{0} &= \cos\frac{\theta}{2}\ket{0}- i\sin\frac{\theta}{2}\ket{1}\\ R_X(\theta)\ket{1} &= - i\sin\frac{\theta}{2}\ket{0}+\cos\frac{\theta}{2}\ket{1}\end{aligned}$$
    $$R_x(\theta) =RXGate(\theta)= e^{-i\frac{\theta}{2}X} = \cos\frac{\theta}{2}\cdot I - i\sin\frac{\theta}{2}\cdot X = \begin{bmatrix}\cos\frac{\theta}{2} & - i\sin\frac{\theta}{2}\\- i\sin\frac{\theta}{2} & \cos\frac{\theta}{2}\end{bmatrix}$$
* `RYGate (ry)` : Rotation about $y$-axis
    $$\begin{aligned}R_X(\theta)\ket{0} &= \cos\frac{\theta}{2}\ket{0}+\sin\frac{\theta}{2}\ket{1}\\ R_X(\theta)\ket{1} &= - \sin\frac{\theta}{2}\ket{0}+\cos\frac{\theta}{2}\ket{1}\end{aligned}$$
    $$R_y(\theta) =RYGate(\theta)= e^{-i\frac{\theta}{2}Y} = \cos\frac{\theta}{2}\cdot I - i\sin\frac{\theta}{2}\cdot Y = \begin{bmatrix}\cos\frac{\theta}{2} & -\sin\frac{\theta}{2}\\ \sin\frac{\theta}{2} & \cos\frac{\theta}{2}\end{bmatrix}$$
* `PhaseGate (p)` : Rotation about about $z$-axis
    $$R_z(\theta) = PhaseGate(\theta)= e^{-i\frac{\theta}{2}Z} = \cos\frac{\theta}{2}\cdot I - i\sin\frac{\theta}{2}\cdot Z = \begin{bmatrix}e^{-i\frac{\theta}{2}} & 0\\0 & e^{i\frac{\theta}{2}}\end{bmatrix}$$
* `RVGate (rv)` : Rotation about arbitrary axis $\vec{v} = \theta\hat{n}$
    $$R_{\hat{n}}(\theta) = RVGate(\theta)= e^{-i\frac{\theta}{2}\hat{n}\cdot\vec{\sigma}} = \cos\frac{\theta}{2}\cdot I - i\sin\frac{\theta}{2}\cdot (\hat{n}\cdot\vec{\sigma})$$
* Any single qubit unitary gate can be written as  $$U = e^{i\alpha}R_{\hat{n}}(\theta)$$

```python
from qiskit.circuit.library import XGate, HGate, RXGate, RYGate, RVGate
from qiskit.visualization import array_to_latex
import numpy as np

X = XGate()
H = HGate()
RX = RXGate(np.pi/2)
RY = RYGate(np.pi/2)
RV = RVGate(1, 1, 1) # v = v_x, v_y, v_z 

# For display only
display(array_to_latex(X.to_matrix()))
display(array_to_latex(H.to_matrix()))
display(array_to_latex(RX.to_matrix()))
display(array_to_latex(RY.to_matrix()))
display(array_to_latex(RV.to_matrix()))
```

## Unitary gate decompositions
* **Z-Y-Z decomposition** $$U = e^{\alpha}R_z(\varphi)R_y(\theta)R_z(\lambda)$$
    * In qiskit, $$U(\theta,\varphi,\lambda)=\begin{bmatrix}\cos\frac{\theta}{2} & -e^{i\lambda}\sin\frac{\theta}{2}\\e^{i\varphi}\sin\frac{\theta}{2} & e^{i(\varphi+\lambda)}\cos\frac{\theta}{2}\end{bmatrix}$$

```python
from qiskit.circuit.library import UGate
import numpy as np
# Create a general gate with specific Euler angles
# theta=pi/2, phi=0, lambda=pi corresponds to a Hadamard-like operation
u = UGate(theta=np.pi/2, phi=0, lam=np.pi)
array_to_latex(u.to_matrix())
```

# Multiple qubit gates
* Tensor product of gates is used to apply single qubit gates to each qubit. $$X\otimes Z (\ket{q_1}\otimes\ket{q_0}) = (X\ket{q_1})\otimes(Z\ket{q_0})$$
* Tensor product of two matrices is a $4\times 4$ matrix.
$$A\otimes B = \begin{bmatrix}a_{11} B & a_{12} B\\ a_{21} B & a_{22} B\end{bmatrix}$$
* In the outer product notation $$\outpro{00}{10} = \outpro{0}{1}\otimes\outpro{0}{0}$$
* The outer products $$\outpro{i}{j}\otimes\outpro{l}{m}$$ form basis for all operators.

## CNOT gate
* Controlled-NOT gate acts on two qubits such that first qubit controls whether NOT gate is applied to the second qubit
    * When the first qubit is in $|0\rangle$, the second qubit remains unchanged.
    * When the first qubit is in $|1\rangle$, the second qubit is flipped.
        $$\begin{align*}
        |0\rangle|0\rangle &\xrightarrow{CNOT} |0\rangle|0\rangle\\
        |0\rangle|1\rangle &\xrightarrow{CNOT} |0\rangle|1\rangle\\
        |1\rangle|0\rangle &\xrightarrow{CNOT} |1\rangle|1\rangle\\
        |1\rangle|1\rangle &\xrightarrow{CNOT} |1\rangle|0\rangle
        \end{align*}$$
        $$CNOT = |0\rangle\langle 0|\otimes I + |1\rangle\langle 1|\otimes X$$
* In qiskit, the order of qubits is reversed.
        $$CNOT = I\otimes |0\rangle\langle 0|+ X\otimes |1\rangle\langle 1|$$

```python
from qiskit.circuit.library import CXGate
from qiskit.visualization import array_to_latex

cnot = CXGate()
array_to_latex(cnot.to_matrix())
```

## Controlled Gate
* For any gate $U$, its controlled version is given by
$$CU = I\otimes |0\rangle \langle 0| + U\otimes |1\rangle\langle 1|$$

* Use `u.control(parameters)` to get controlled version of `u`, where the parameters are
    * `name` -- `string`, name of the gate
    * `num_qubits` -- `int`, number of qubits the gate acts on
    * `label` -- optional label for the gate like `CU`
    * `num_ctrl_qubits` -- `int`, number of control qubits (default 1)
    * `ctrl_state` -- `int`, `string`, control state in decimal or bit string

```python
from qiskit.quantum_info import Operator
from qiskit.circuit.library import HGate
from qiskit.visualization import array_to_latex
# Two qubit control gate, by default 0 qubit is control, 1 qubit is target
CH = HGate().control(num_ctrl_qubits=2, ctrl_state="11")
CH_op = Operator(CH)
array_to_latex(CH_op.data)
```

# Universality
* Every unitary operator can be decomposed in terms of CNOT and single qubit gates.
* Every gate $U$ can be expressed as $$U = S_nC_nS_{n-1}C_{n-1}\cdots S_0C_0$$
    where $S_i$ are product of single qubit gates and $C_i$ are CNOT applied to chosen two qubits.

# Mapping input to states
* **Number** : convert to binary to quantum state
    $$17 \to 10001 \to \ket{10001}$$
* **String** : convert characters to numbers and then to states
    $$"ABC" \to 65~66~67\to 1000001~1000010~100011\to \ket{1000001~1000010~100011}$$
* **List** : fix number of bits for each element and then to states
* **Graph** : number vertices, vertex pair for edges
* **Functions** : create a unitary gate such that $$U_f\ket{x}\ket{y} = \ket{x}\ket{y\oplus f(x)}$$

# Utilizing superposition
* Create superposition of all possible inputs $$\frac{1}{\sqrt{N}}\sum_x\ket{x}$$
* Apply $U_f$ on superposition $$U_f(\frac{1}{\sqrt{N}}\sum_x\ket{x})\ket{0} = \frac{1}{\sqrt{N}}\sum_x\ket{x}\ket{f(x)}$$
    * $f$ has been evaluated on all inputs in one application of $U_f$ only!!
    * This is **quantum parallelism**.
