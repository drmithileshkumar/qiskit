# Hiden subgroup problem
* Domain: A group $G$
* Range: A set $X$
* Subgroup: Hidden subgroup $K$ 
* A function $F: G\to X$ such that $F$ is constant on cosets of $K$
## Deutsch Problem
* Input: $f:\{0,1\}\to\{0,1\}$ which can be either constant or balanced.
* Task: Determise whether $f$ is constant or balanced.
* $G=\{0,1\}$ with addition modulo 2 operations
* $X=\{0,1\}$
* $K = \{0,1\}$ for constant and $K = \{0\}$ for balanced
* $F$ is constant on cosets of $K$
## Deutsch-Jozsa Problem
* Input: $f:\{0,1\}^n\to\{0,1\}$ which can be either constant or balanced.
* Task: Determise whether $f$ is constant or balanced.
* $G=\{0,1\}^n$ (with co-ordinate-wise $\oplus_p$ on permutated set)
* $X=\{0,1\}$
* $K = \{0,1\}^n$ for constant and $K = A\subset{\{0,1\}^n}$ such that $f(A) = 0$ for balanced
* $F$ is constant on cosets of $K$
* The idea is to map $A\to\{0\}\times \{0,1\}^{n-1}$ and $B\to\{1\}\times\{0,1\}^{n-1}$
* The above mapping $p$ can be chosen in lexographic order in $A$ and $B$.
                    $$x\oplus_p y = p^{-1}(p(x)\oplus^n p(y))$$
                    where $\oplus^n$ is bit-wise $\oplus$.
## Simon's algorithm
* Input: $f:\{0,1\}^n\to\{0,1\}^n$
* Promise: $f$ is one-to-one or two-to-one
* There is $s$ such that $f(x \oplus s) = f(x)$
    * In case of one-to-one: $s = 0^n$
    * In case of two-to-one: $s$ is unique non-zero string
* Task: Find $s$
* $G = \{0,1\}^n$ with bit-wise $\oplus$ operation
* $X = \{0,1\}^n$
* $F = f: f(x\oplus s) = f(x)$
* $K = \{0^n\}$ or $K = \{0^n, s\}$
## Period-finding
* Input: $f: Z\to Z$ such that $\exists r$ $f(x + r) = f(x)$
* Task: find $r$
* $G= Z$ with operation $+$
* $X = Z$
* $K = \{0, r, 2r, \dots\}$
* $F : f(x + r) = f(x)$
## Order-finding
* Input: $N$ and $a$ such that $1\leq a < N$ is co-prime to $N$
* Task: Find smallest $r$ such that $a^r\equiv 1\mod N$
* $G = Z, +$
* $X = \{1, a, a^2 \dots, a^{r-1}\}$
* $K = \{0, r, 2r, \dots\}$
* $f(x) = a^x\mod N$ such that $f(x + r) = f(x)$
