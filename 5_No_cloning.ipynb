# Qubit copying gate
* Does there exist a unitary transformation such that $|a\rangle|0\rangle\to|a\rangle|a\rangle$?
* A first attempt: Use $|a\rangle$ as control, i.e. use CNOT gate $$\begin{align*}|0\rangle|0\rangle&\xrightarrow{CNOT}|0\rangle|0\rangle\\ |1\rangle|0\rangle&\xrightarrow{CNOT}|1\rangle|1\rangle\end{align*}$$
    * This seems to work. Lets apply it to a general state $|\psi\rangle = \alpha |0\rangle + \beta |1\rangle$, $$|\psi\rangle|0\rangle = (\alpha|0\rangle + \beta|1\rangle)|0\rangle = \alpha|0\rangle|0\rangle + \beta|1\rangle|0\rangle \xrightarrow{CNOT}\alpha |0\rangle|0\rangle + \beta |1\rangle|1\rangle\neq |\psi\rangle|\psi\rangle$$
* May be using only CNOT is not a good idea.
* Can some other unitary transform work?
    * No-cloning theorem says no, its impossible.

# No-cloning theorem
* **Input** : Unknown single qubit state $|\psi\rangle$ and known ancilla qubits
* **Cloning Objective** : Find a unitary transformation $U$ such that $$|\psi\rangle\otimes |u\rangle\xrightarrow{U}|\psi\rangle\otimes|\psi\rangle$$
* **Theorem** : $U$ does not exist.
* ***Proof*** :
    * Start with two states $|a\rangle$ and $|b\rangle$ that are neither parallel nor orthogonal and need to be cloned.
                        $$
                        \begin{align*}
                        U(|a\rangle|u\rangle) &= |a\rangle|a\rangle\\
                        U(|b\rangle|u\rangle) &= |b\rangle|b\rangle
                        \end{align*}
                        $$
    * Consider their inner product 
                        $$
                        \begin{align*}
                        \langle au|U^\dagger U|bu\rangle &= \langle aa|bb\rangle\\
                        \langle a|b\rangle &= \langle a|b\rangle^2\\
                        \implies \langle a|b\rangle &= 0\text{ or } 1
                        \end{align*}
                        $$
    * This implies that either $|a\rangle$ and $|b\rangle$ are parallel or orthogonal to each other.
    * Hence, a univarsal cloning unitary operator $U$ does not exist.

## How close can we come to cloning?
* So $\psi \to \psi$ cloning is not possible, try approximate cloning, i.e. $\rho\to \sigma$ cloning where $\sigma$ is as close to $\rho$ as possible.
* Consider cloning unitary as follows: $$
\begin{aligned}
    \ket{0_{in}}\ket{0_{target}}\ket{0_{work}} &\to a \ket{00}\ket{0} + b \ket{\beta_{01}}\ket{1}\\
    \ket{1_{in}}\ket{0_{target}}\ket{0_{work}} &\to a \ket{11}\ket{1} + b \ket{\beta_{01}}\ket{0}
\end{aligned}$$
* Trace out work and input qubits and compare the reduced density matrix to $\rho = \outpro{\psi}{\psi}$
* Best cloning achievable applies the following transformation $$|\psi\rangle\langle\psi|\xrightarrow{U_c}\frac{2}{3}|\psi\rangle\langle\psi|+\frac{1}{3}\frac{I}{2}$$
* The fidelity of this state is $$\providecommand{\expect}[3]{\langle #1\vert #2\vert #3\rangle} F = \expect{\psi}{\rho}{\psi}=\frac{5}{6}$$

## Implication
* Any algorithm that requires cloning or copying an intermediate state will not work in quantum computing.

### Are there other such impossible operations we need to be awared of?
* Recall that classical circuits used swapping as well.
    * We have already lost universal cloning.
    * Do we lose universal swap as well?
* It turns out universal swap works.
