## Problem 1: Simulating Synthetic "Fractional" Matter

---

### 1. Context: The Rules of the Universe
In our universe, nature follows a strict "menu" called the **Periodic Table**. Each element is defined by its **Nuclear Charge ($Z$)**, which is always a whole number. For example, Hydrogen has a charge of **+1** (one proton), and Helium has a charge of **+2**.

This charge acts like a magnet that pulls electrons toward the center. The strength of this pull determines the "glue" of our world: how big an atom is, how strongly it bonds to others, and whether a molecule is stable or falls apart. 

In a chemistry lab, you cannot create an atom with a charge of **+0.5**. However, in a **Quantum Simulator**, the physics of a molecule is defined by a mathematical operator called a **Hamiltonian**. Since you are the programmer, you can "hack" the laws of physics. You can build a synthetic universe where protons are only half as strong as ours and discover if "matter" could even exist there.

---

### 2. The Objective
Your goal is to use **Qiskit** to determine the **Critical Stability Point** of a synthetic diatomic molecule (a molecule made of two identical "custom" atoms). 

You will simulate a series of "Fractional Hydrogen" molecules where the nuclear charge $Z$ is treated as a variable. You must find the **minimum charge ($Z_{crit}$)** required to keep two atoms bonded together.

---

### 3. Your Tasks
1.  **Construct the Synthetic Hamiltonian:** Create a script that generates a molecular Hamiltonian where the nuclear charge $Z$ is a floating-point variable (e.g., $Z = 0.4, 0.7, 1.2$).
2.  **Map the Energy Landscapes:** For different values of $Z$, calculate the ground state energy at various inter-atomic distances (from 0.5Å to 4.0Å) using the **VQE (Variational Quantum Eigensolver)** algorithm.
3.  **Identify the Stability Threshold:**
    * If your energy-vs-distance plot shows a "dip" (a potential well), the molecule is **stable**.
    * If the plot is a flat line or always increases, the atoms refuse to bond and the molecule is **impossible**.
    * Find the specific value of $Z$ where the "dip" disappears.
4.  **Analyze the "Bloat" Effect:** Document how the equilibrium bond distance (the distance where the energy is lowest) changes as the charge $Z$ weakens.

---

### 4. Technical Constraints
* **Software:** Use `qiskit-nature` (v0.7+) and `qiskit-algorithms`.
* **Basis Set:** To keep simulation times low, use the **STO-3G** basis set.
* **Reference:** Start by successfully simulating a standard Hydrogen molecule ($Z=1.0$) to verify your VQE setup before moving to fractional charges.

---

### 5. Deliverables
Submit a Python Notebook (`.ipynb`) containing:
* A clear explanation (with code comments) of how you modified the Qiskit Hamiltonian to account for fractional nuclear charges.
* A single plot containing multiple **Potential Energy Curves** for different values of $Z$.
* Your final estimate for $Z_{crit}$ (the "Limit of Chemistry").
* **Discussion Question:** If you were an inhabitant of a $Z=0.6$ universe, would your "water" molecules be larger or smaller than ours? Explain your reasoning based on your simulation data.

## Problem 2: Forensic Fingerprinting of Quantum Hardware

---

### 1. Context: The Digital Fingerprint
In the classical world, no two pieces of hardware are truly identical. Every digital camera sensor has "hot pixels," and every printer has microscopic patterns in how it sprays ink. These imperfections act as a **fingerprint** that can be used to identify which specific machine created a document or a photo.

Quantum computers are no different. In the current **NISQ (Noisy Intermediate-Scale Quantum)** era, every Quantum Processing Unit (QPU) has its own "personality." This personality is defined by its unique noise profile: how often a bit flips accidentally, how quickly a qubit loses its quantum state ($T_1$ and $T_2$ times), and how much "crosstalk" exists between neighbors. 

If you run the same circuit on two different IBM quantum computers, the results will look slightly different because of these "ghosts" in the hardware.

---

### 2. The Objective
Your goal is to act as a **Quantum Forensic Analyst**. You will build a "Fingerprint Library" for several simulated quantum backends and then write a classification script that can identify an "Unknown" backend based solely on the noise in its output.

---

### 3. Your Tasks
1.  **Characterize the Targets:** Use `qiskit-aer` to create three distinct noisy simulators (using `NoiseModel.from_backend`) based on real IBM devices (e.g., `ibm_brisbane`, `ibm_kyoto`, and `ibm_osaka`).
2.  **Design "Probe" Circuits:** Create a suite of 3–5 short circuits designed to "stress" the hardware and amplify its specific noise. 
    * *Examples:* Long chains of Identity gates to measure decay, or simultaneous $X$ gates to measure readout error.
3.  **Generate the Library:** Run your probes on each noisy simulator and store the resulting probability distributions (the "fingerprints") in a dictionary or JSON file.
4.  **The Blind Test:** * I (or a script) will provide a set of "Unknown" measurement counts. 
    * You must write a function that compares the "Unknown" counts against your library using a statistical distance metric (like **Hellinger Distance** or **Kullback-Leibler Divergence**).
5.  **Identify the Source:** Correctively label which of your three target backends produced the unknown data.

---

### 4. Technical Constraints
* **Software:** Use `qiskit` (v1.0+) and `qiskit-aer`.
* **Metrics:** You must use a formal statistical method to compare distributions; "eyeballing" the graphs is not sufficient for full credit.
* **Consistency:** Ensure you use the same number of `shots` for both your library generation and your blind test to maintain statistical parity.

---

### 5. Deliverables
Submit a Python Notebook (`.ipynb`) containing:
* Your **Fingerprint Library** showing the unique noise signatures of your chosen backends.
* The code for your **Classification Engine**.
* A **Confusion Matrix** or success report showing how accurately your engine can distinguish between the backends.
* **Discussion Question:** Could a "hacker" intentionally add noise to their circuit to hide which quantum computer they were using? How would that affect the fidelity of their actual calculation?

## Problem 3: Flexible Representation of Quantum Images (FRQI)

---

### 1. Context: Pixels vs. Amplitudes
In a classical computer, an image is stored as a massive grid of numbers (RGB values). For a high-resolution photo, this requires millions of bits. In the quantum world, we can use a property called **Superposition** to store information much more densely.

The **FRQI (Flexible Representation of Quantum Images)** method is a fundamental technique for encoding grayscale images into a quantum state. Instead of using one qubit per pixel, FRQI uses the **probability amplitudes** of a quantum state to represent the intensity (brightness) of pixels. This means you can theoretically store a 4-pixel image using only 2 "position" qubits and 1 "color" qubit.

---

### 2. The Objective
Your goal is to build a Quantum Image Encoder and Decoder. You will take a $2 \times 2$ grayscale image (4 pixels), "compress" it into a 3-qubit quantum circuit, and then run a simulation to "retrieve" the image by measuring the quantum state.

---

### 3. Your Tasks
1.  **Prepare the Image Data:** Define a $2 \times 2$ matrix where each value represents a grayscale intensity between 0 (black) and 1 (white).
2.  **Construct the Position State:** Use $H$ gates to put your "position" qubits into a uniform superposition so that all 4 pixel locations $(|00\rangle, |01\rangle, |10\rangle, |11\rangle)$ exist simultaneously.
3.  **The FRQI Encoding:** Implement a series of **Multi-Controlled $R_y$ rotations**. 
    * Each rotation should target the "color" qubit but be conditioned on a specific "position" qubit state.
    * The angle of rotation $\theta$ will represent the brightness of that specific pixel.
4.  **The Measurement Decoder:** Since you cannot "see" a quantum state directly, you must run the circuit for a high number of `shots` (e.g., 8192). 
    * Calculate the probability of measuring a $|1\rangle$ on the color qubit for each of the 4 position states.
    * Map these probabilities back to grayscale values to reconstruct the image.

---

### 4. Technical Constraints
* **Software:** Use `qiskit` (v1.0+) and `matplotlib` for visualization.
* **Math:** Remember that in FRQI, the probability of measuring $|1\rangle$ is proportional to $\cos^2(\theta)$. You will need to apply the inverse math to get your original pixel values back.
* **Accuracy:** Compare your **Original Image** to your **Reconstructed Image**. Calculate the "Root Mean Square Error" (RMSE) to see how much information was lost due to finite sampling (shots).

---

### 5. Deliverables
Submit a Python Notebook (`.ipynb`) containing:
* The circuit diagram for your $2 \times 2$ FRQI encoder.
* A side-by-side visualization showing the **Original 4-pixel grid** vs. the **Reconstructed 4-pixel grid**.
* A plot showing how the reconstruction error (RMSE) decreases as you increase the number of `shots`.
* **Discussion Question:** If you wanted to encode an $8 \times 8$ image (64 pixels), how many total qubits would you need? Why is this considered "exponentially" more efficient than classical storage?

## Problem 4: Portfolio Optimization with Real Market Data

---

### 1. Context: The Risk-Reward Balance
In the financial world, an investor’s goal is to pick a "Portfolio" (a collection of stocks) that provides the highest possible return for the lowest possible risk. This is known as **Mean-Variance Optimization**. 

As the number of available stocks increases, the number of possible combinations grows exponentially. Finding the "Perfect" balance becomes a massive search problem. While classical computers use calculus to solve this, Quantum Computers use an algorithm called **QAOA (Quantum Approximate Optimization Algorithm)** to "shape" a quantum state until it represents the most profitable combination of assets.

---

### 2. The Objective
Your goal is to build a hybrid quantum-classical pipeline that pulls **real-time stock market data**, processes it into a mathematical format a quantum computer understands, and uses a quantum algorithm to suggest the best 3-stock portfolio from a list of 6 candidates.

---

### 3. Your Tasks

2.  **Calculate the Statistics:** Compute the **Expected Returns** (how much the stock goes up on average) and the **Covariance Matrix** (how much the stocks move together or apart).
3.  **Formulate the QUBO:** Translate these financial statistics into a **Quadratic Unconstrained Binary Optimization (QUBO)** problem.
    * *Constraint:* You must set a "Budget" of exactly 3 stocks. The algorithm should penalize any solution that picks more or fewer than 3.
4.  **The Quantum Solver:** Use `qiskit_optimization` and the `SamplingVQE` or `QAOA` algorithm to solve the problem.
5.  **Benchmark:** Solve the exact same problem using a classical solver (like `NumPyMinimumEigensolver` or a brute-force search) and compare the results.

---

### 4. Technical Constraints
* **Software:** Use `qiskit-optimization` and `yfinance`.
* **Penalty Factor:** You will need to choose a "Penalty" value ($\rho$) for your budget constraint. If it’s too low, the quantum computer will ignore your budget; if it’s too high, it won't find the best stocks.
* **Backend:** You may use the `AerSimulator` for your quantum runs.

---

### 5. Deliverables
Submit a Python Notebook (`.ipynb`) containing:
* A table showing the **Expected Returns** and **Risk** of the 6 stocks you chose.
* The **Quantum Circuit** generated by the QAOA algorithm for this specific optimization.
* A final "Investment Recommendation" showing the 3 stocks the quantum computer selected.
* **Discussion Question:** Look at your Covariance Matrix. Did the quantum computer pick stocks that move together (correlated) or stocks that move differently (diversified)? Why is diversification important for reducing risk in a portfolio?

## Problem 5: Quantum Nim (The State-Eater Game)

---

### 1. Context: The Game of Nim
In the classical game of **Nim**, players take turns removing objects (like matches or stones) from distinct heaps. The person who takes the last object wins (or loses, depending on the variation). It is a game of pure mathematical strategy, where every position can be defined as "winning" or "losing" based on binary logic.

In **Quantum Nim**, we don't have heaps of stones. Instead, we have a **shared quantum state** (a multi-qubit system). The "objects" are the possible outcomes (basis states) of the system. Instead of simply picking up a stone, a player's "move" consists of applying a **Unitary Transformation** or a **Partial Measurement** to "eat" or collapse parts of the quantum state.

---

### 2. The Objective
Your goal is to build a 3-qubit Quantum Nim engine. You will start with a highly entangled state (a **W-state**), and two players (Alice and Bob) will take turns "consuming" the entanglement. The winner is the player who successfully forces the system into a specific "Target State" (the $|000\rangle$ ground state) after a set number of moves.

---

### 3. Your Tasks
1.  **Initialize the "Heap":** Write a function to prepare a 3-qubit **W-state**: 
    $$|W\rangle = \frac{1}{\sqrt{3}}(|001\rangle + |010\rangle + |100\rangle)$$
    This state represents three "objects" existing in a quantum superposition.
2.  **Define the Move Set:** Each player can choose one of two legal moves per turn:
    * **The Rotation:** Apply a $Ry(\theta)$ gate to a qubit of their choice to shift the probability toward $|0\rangle$.
    * **The Measure-and-Collapse:** Perform a measurement on one qubit. If it collapses to $|1\rangle$, that "object" is removed from the heap. If $|0\rangle$, the game continues with the remaining superposition.
3.  **The Engine:** Create a game loop where Alice and Bob play against each other.
4.  **Strategic Analysis:** * Find the "Optimal Rotation Angle" ($\theta$) that Alice should use on the first turn to maximize her probability of winning, regardless of what Bob does.
    * Prove whether it is statistically better to **Rotate** or **Measure** when the system is in a state of high entanglement.

---

### 4. Technical Constraints
* **Software:** Use `qiskit` (v1.0+) and the `Statevector` class for precise probability tracking.
* **Winning Condition:** A player wins if, after their move, the probability of the system being in the $|000\rangle$ state is $> 90\%$.
* **Simulation:** Run 500 simulated games of "Random Move" vs "Strategic Move" to collect win-rate data.

---

### 5. Deliverables
Submit a Python Notebook (`.ipynb`) containing:
* The code for the **W-state** preparation and the interactive Game Loop.
* A **Win-Rate Heatmap** showing how different rotation angles $\theta$ affect the outcome for the first player.
* **Discussion Question:** In classical Nim, the game is "Solved" (the first player can always win if they know the math). Is Quantum Nim also "Solved"? Based on your entanglement simulations, does the first player still have a guaranteed advantage, or does quantum uncertainty make the game unpredictable?
