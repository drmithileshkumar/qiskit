# Order-finding algorithm
* Input: Positive integers $N$ and $a \lt N$.
* Task: Find smallest integer $r$ such that $$a^r\equiv 1\mod N$$
## Oracle
* Construct the oracle $U_a$ such that $$U_a|x\rangle = |ax\mod N\rangle$$
* $U_a$ is unitary because it is effectively a permutation.
* Here after, we assume that all multiplications within a ket are $\mod N$.
* What are the eigenvalues and eigenvectors of $U_a$? 
* Notice that $$U_a^r|x\rangle = |a^rx\rangle = |x\rangle$$
### Eigenvalues
* Let $\varphi$ be phase of an eigenvalue and $|u\rangle$ be the corresponding eigenvector. $$U_a|u\rangle = e^{2\pi i\varphi}|u\rangle$$
* Applying $U_a^r$ we get $$\begin{align}
                                U_a^r|u\rangle &= e^{2\pi i r\varphi}|u\rangle = |u\rangle\\ 
                                \implies 2\pi r\varphi &= 2\pi s\\
                                \implies \varphi &= \frac{s}{r} 
                            \end{align}$$
* The only measureable values of $s$ are $0\leq s\leq r-1$.
### Eigenvectors
* Vectors $|a^0\rangle, |a^1\rangle, \dots, |a^{r-1}\rangle$ are orthogonal and span an $r$-dimensional subspace such that 
        $$\begin{align}
            U_a|a^k\rangle &= |a^{k + 1}\rangle\\
            U_a|a^{r-1}\rangle &= |1\rangle
        \end{align}$$
* Claim: $|u_s\rangle = \frac{1}{\sqrt{r}}\sum_{k=0}^{r-1}e^{-2\pi isk/r}|a^k\rangle$.
    * This is easily verifiable as $U_a|u_s\rangle = \frac{1}{\sqrt{r}}\sum_{k=0}^{r-1}e^{-2\pi isk/r}|a^{k+1}\rangle=e^{2\pi is/r}|u_s\rangle$
* Even though we know the form of $|u_s\rangle$ from above, we can not use it as $r$ is not known.
* Claim: $\frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}|u_s\rangle = |1\rangle$
    * $$\begin{align}
            \frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}|u_s\rangle &= \frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}\frac{1}{\sqrt{r}}\sum_{k=0}^{r-1}e^{-2\pi isk/r}|a^k\rangle\\
            &= \frac{1}{r}\sum_{k,s}e^{-2\pi isk/r}|a^k\rangle\\
            &= \frac{1}{r}\sum_k(\sum_s (e^{-2\pi ik/r})^s)|a^k\rangle\\
            &= \frac{1}{r}\sum_k r\delta_{0,k}|a^k\rangle\\
            &= |1\rangle
        \end{align}$$
* Starting with $|1\rangle$ instead of $|u_s\rangle$, we can use phase estimation algorithm.
# Algorithm
$$\begin{align}
|0^{\otimes t}\rangle|1\rangle &\xrightarrow{H^{\otimes t}} \frac{1}{\sqrt{2^t}}\sum_{j = 0}^{2^t - 1}|j\rangle |1\rangle\\
&\xrightarrow{c-U^j_a} \frac{1}{\sqrt{2^t}}\sum_{j = 0}^{2^t - 1}|j\rangle |a^j\rangle\\
&= \frac{1}{\sqrt{r2^t}}\sum_{s=0}^{r-1}\sum_{j = 0}^{2^t - 1}e^{2\pi r sj/r}|j\rangle |u_s\rangle\\
&\xrightarrow{QFT^\dagger}\frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}|\varphi_{s,t}\rangle |u_s\rangle\\
&\xrightarrow{measure}\varphi_{s,t}\quad (t-\text{bits of } \frac{s}{r})\\
&\xrightarrow{continued fraction} r
\end{align}$$
