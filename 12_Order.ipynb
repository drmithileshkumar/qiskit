# Order-finding algorithm
* **Input**: Positive integers $N$ and $a \lt N$ and they are coprime to each other.
* **Task**: Find smallest integer $r$ such that $$a^r\equiv 1\bmod N$$
## Oracle
* Construct the oracle $U_a$ such that $$U_a|x\rangle = |ax\bmod N\rangle$$
* Here after, we assume that all multiplications within a ket are $\bmod N$.
* $U_a$ is unitary because it is effectively a permutation.
    * If $a$ and $N$ are not coprime, then let $d = gcd(N,a)$.
    * For every $x$, there is $x + N/a$ an integer.
    * $U_a\ket{x + N/a} = \ket{ax + N} = \ket{ax} = U_a\ket{x}$, hence $U_a$ is not unitary.

* What are the eigenvalues and eigenvectors of $U_a$? 
* Notice that $$U_a^r|x\rangle = |a^rx\rangle = |x\rangle$$
### Eigenvalues
* Let $\varphi$ be phase of an eigenvalue and $|u\rangle$ be the corresponding eigenvector. $$U_a|u\rangle = e^{2\pi i\varphi}|u\rangle$$
* Applying $U_a^r$ we get $$\begin{align}
                                U_a^r|u\rangle &= e^{2\pi i r\varphi}|u\rangle = |u\rangle\\ 
                                \implies 2\pi r\varphi &= 2\pi s\\
                                \implies \varphi &= \frac{s}{r} 
                            \end{align}$$
* The only measureable values of $s$ are $0\leq s\leq r-1$.
### Eigenvectors
* Vectors $|a^0\rangle, |a^1\rangle, \dots, |a^{r-1}\rangle$ are orthogonal and span an $r$-dimensional subspace such that 
        $$\begin{align}
            U_a|a^k\rangle &= |a^{k + 1}\rangle\\
            U_a|a^{r-1}\rangle &= |1\rangle
        \end{align}$$
* **Claim**: $|u_s\rangle = \frac{1}{\sqrt{r}}\sum_{k=0}^{r-1}e^{-2\pi isk/r}|a^k\rangle$.
    * This is easily verifiable as $U_a|u_s\rangle = \frac{1}{\sqrt{r}}\sum_{k=0}^{r-1}e^{-2\pi isk/r}|a^{k+1}\rangle=e^{2\pi is/r}|u_s\rangle$
* Even though we know the form of $|u_s\rangle$ from above, we can not use it as $r$ is not known.
* **Claim**: $\frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}|u_s\rangle = |1\rangle$
    * $$\begin{align}
            \frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}|u_s\rangle &= \frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}\frac{1}{\sqrt{r}}\sum_{k=0}^{r-1}e^{-2\pi isk/r}|a^k\rangle\\
            &= \frac{1}{r}\sum_{k,s}e^{-2\pi isk/r}|a^k\rangle\\
            &= \frac{1}{r}\sum_k(\sum_s (e^{-2\pi ik/r})^s)|a^k\rangle\\
            &= \frac{1}{r}\sum_k r\delta_{0,k}|a^k\rangle\\
            &= |1\rangle
        \end{align}$$
* Starting with $|1\rangle$ instead of $|u_s\rangle$, we can use phase estimation algorithm.
# Algorithm
$$\begin{align}
|0^{\otimes t}\rangle|1\rangle &\xrightarrow{H^{\otimes t}} \frac{1}{\sqrt{2^t}}\sum_{j = 0}^{2^t - 1}|j\rangle |1\rangle\\
&\xrightarrow{c-U^j_a} \frac{1}{\sqrt{2^t}}\sum_{j = 0}^{2^t - 1}|j\rangle |a^j\rangle\\
&= \frac{1}{\sqrt{r2^t}}\sum_{s=0}^{r-1}\sum_{j = 0}^{2^t - 1}e^{2\pi r sj/r}|j\rangle |u_s\rangle\\
&\xrightarrow{QFT^\dagger}\frac{1}{\sqrt{r}}\sum_{s=0}^{r-1}|\varphi_{s,t}\rangle |u_s\rangle\\
&\xrightarrow{measure}\varphi_{s,t}\quad (t-\text{bits of } \frac{s}{r})\\
&\xrightarrow{continued fraction} r
\end{align}$$


# Continued fractions
* Express a real number as $x = [a_0, a_1, \dots, a_N]$:
$$\begin{aligned}
x &= a_0 + \frac{1}{a_1 + \frac{1}{a_2 + \frac{1}{\cdots + \frac{1}{a_N}}}}\end{aligned}$$
$$\begin{aligned}
\frac{22}{7} &= 3 + \frac{1}{7} = [3, 7]\\
\frac{31}{13} &= 2 + \frac{5}{13} = 2 + \frac{1}{\frac{13}{5}} = 2 + \frac{1}{2 + \frac{3}{5}}\\
              &= 2 + \frac{1}{2 + \frac{1}{\frac{5}{3}}} = 2 + \frac{1}{2 + \frac{1}{1 + \frac{2}{3}}}= 2 + \frac{1}{2 + \frac{1}{1 + \frac{1}{\frac{3}{2}}}}\\
              &= 2 + \frac{1}{2 + \frac{1}{1 + \frac{1}{1 + \frac{1}{2}}}} = [2, 2, 1, 1, 2]
\end{aligned}
$$
* If $x = [a_0, a_1, \dots, a_N]$, then $[a_0, \dots, a_i] = \frac{p_i}{q_i}$ are called convergents of $x$.
* Recurrence relation for a sequence of positive numbers $a_0, a_1, \dots, a_N$ defining the continued fractions $$[a_0, \dots, a_n] = \frac{p_n}{q_n}$$ is given by 
$$ \begin{aligned}
p_0 &= a_0\quad q_0 = 1\\
p_1 &= 1 + a_0a_1\quad q_1 = a_1\\
p_n &= a_n p_{n-1} + p_{n-2}\\
q_n &= a_n q_{n-1} + q_{n-2}
\end{aligned}
$$
* **Theorem**: If $p/q$ is within $1/2q^2$ of a rational number $x$, then it is a convergent of continued fraction for $x$.
* Application to Order finding:
    * $\varphi_{s,t}$ is within $1/2^{2t+1}$ of desired $s/r$. As $r\leq N\leq 2^t$, above theorem applies.

# Factoring
* Task: Find factors of an integer $N$.
* Choose a number $x$ between $3$ and $N-1$. 
    * If $gcd(x, N) > 1$, then we have found a factor.
    * Otherwise, $gcd(x, N) = 1$. Find order $r$ of $x\bmod N$.
    * If $r$ is even with $x^{r/2}\neq -1\bmod N$,
        * Compute $gcd(x^{r/2} - 1, N)$ and $gcd(x^{r/2} + 1, N)$
        * One of them is likely a factor.
        * Otherwise repeat, i.e. start with another $x$.
* **Theorem**: If $a^2 = 1\bmod N$ such that $a\neq \pm 1\bmod N$, then at least one of $gcd(a\pm 1, N)$ is a factor of $N$.
    * Trivially, $N$ divides $a^2-1=(a+1)(a-1)$. Hence, both $a\pm 1$ can not be coprime to $N$ at the same time.
